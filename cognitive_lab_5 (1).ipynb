```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print("Original array:", arr)
print("After adding 2:", arr + 2)
print("After multiplying by 3:", arr * 3)
print("After dividing by 2:", arr / 2)
```

```python

arr = np.array([1, 2, 3, 6, 4, 5])
print("Original array:", arr)
reverse_arr = arr[::-1]
print("Reversed array:", reverse_arr)
```

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
max_index = np.argmax(counts)
most_frequent = values[max_index]
indices = np.where(x == most_frequent)[0]
print("Array:", x)
print("Most frequent value:", most_frequent)
print("Frequency:", counts[max_index])
print("Indices:", indices)
```

```python

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)

max_count = np.max(counts)

most_frequent_values = values[counts == max_count]

print("Most frequent values:", most_frequent_values)
print("Frequency:", max_count)

for value in most_frequent_values:
    indices = np.where(y == value)[0]
    print("Value:", value, "Indices:", indices)
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])
```

```python
sparsh = np.linspace(10, 100, 25)
print("Array:")
print(sparsh)
print("Dimensions:", sparsh.ndim)
print("Shape:", sparsh.shape)
print("Total elements:", sparsh.size)
print("Data type:", sparsh.dtype)
print("Total bytes:", sparsh.nbytes)
transpose_sparsh = sparsh.reshape(25, 1)
print("Transpose using reshape():")
print(transpose_sparsh)
print("Transpose using T:")
print(sparsh.T)
```

```python
ucs420_sparsh = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])
print("Original Array:")
print(ucs420_sparsh)
print("Mean:", np.mean(ucs420_sparsh))
print("Median:", np.median(ucs420_sparsh))
print("Maximum:", np.max(ucs420_sparsh))
print("Minimum:", np.min(ucs420_sparsh))
print("Unique elements:", np.unique(ucs420_sparsh))
reshaped_ucs420_sparsh = ucs420_sparsh.reshape(4, 3)
print("\nReshaped Array:")
print(reshaped_ucs420_sparsh)
resized_ucs420_sparsh = np.resize(ucs420_sparsh, (2, 3))
print("\nResized Array:")
print(resized_ucs420_sparsh)
```
