```python
n = int(input("Enter a number: "))

sum = 0

for i in range(2, n + 1):
    prime = True

    for j in range(2, i):
        if i % j == 0:
            prime = False
            break

    if prime:
        sum = sum + i

print("Sum of prime numbers =", sum)
```

```python
def add_odd(n):
    sum = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            sum = sum + i

    return sum


n = int(input("Enter a number: "))
print("Sum of odd numbers =", add_odd(n))
```

```python
def add_prime(n):
    sum = 0

    for i in range(2, n + 1):
        prime = True

        for j in range(2, i):
            if i % j == 0:
                prime = False
                break

        if prime:
            sum = sum + i

    return sum


n = int(input("Enter a number: "))
print("Sum of prime numbers =", add_prime(n))
```
